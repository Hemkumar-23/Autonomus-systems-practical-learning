```python
import tensorflow as tf

print("TensorFlow version:", tf.__version__)

temperature_c = tf.constant(25.0)
temperature_f = temperature_c * 1.8 + 32

print("Temperature in Fahrenheit:", temperature_f.numpy())
```

```python
import tensorflow as tf

# Five temperature readings from a motor
readings = tf.constant([25.0, 27.0, 29.0, 28.0, 31.0])

print("Readings:", readings.numpy()) #Holds all five temperature measurements together in one TensorFlow tensor.
print("Shape:", readings.shape) #There is one dimension containing five elements. The comma indicates a one-element shape tuple in Python.
#A tensor's rank tells us how many dimensions it has:
#Rank 0: one scalar value, such as 25.
#Rank 1: a sequence of values, such as [25, 27, 29, 28, 31].
#Rank 2: a table of values, such as readings from several sensors over time.
#Rank 3 or higher: multidimensional data, including representations of images and batches of images.

print("Rank:", tf.rank(readings).numpy())
print("First reading:", readings[0].numpy())
```

```python
corrected_readings = readings + 2.0 #no for loop is used, by default each tensor value elements is added with the number
#note: TensorFlow supports element-wise operations on tensors
print("Original:", readings.numpy())
print("Corrected:", corrected_readings.numpy())
```
